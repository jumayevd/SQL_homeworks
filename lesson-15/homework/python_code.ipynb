```python
import pyodbc
import json
```

```python
conn = pyodbc.connect('DRIVER={SQL Server};SERVER=ASUS-VivoBook\DEV_EDITION;DATABASE=lesson15;Trusted_Connection=yes;')
cursor = conn.cursor()

# SQL Query to Retrieve JSON Data
sql_query = """
WITH ContactProperties AS (
    SELECT 
        identifier_value AS vid,
        (
            SELECT 
                firstname AS property, firstname AS value 
                FOR JSON PATH
        ) AS properties
    FROM Contacts
)
SELECT vid, properties 
FROM ContactProperties
FOR JSON PATH, ROOT('contacts');
"""

cursor.execute(sql_query)
json_data = cursor.fetchone()[0]  # Fetch the JSON result

cursor.close()
conn.close()

# Print JSON Output (Formatted)
parsed_json = json.loads(json_data)
print(json.dumps(parsed_json, indent=4))
```
